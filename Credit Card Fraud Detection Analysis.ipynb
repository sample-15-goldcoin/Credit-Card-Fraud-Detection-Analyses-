```python
Q1 = df["amt"].quantile(0.25)
Q3 = df["amt"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
plt.figure(figsize=(10, 5))

plt.boxplot(df["amt"], vert=False)

plt.title("Outlier Detection in Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.show()
```

```python
df["amount_outlier"] = (
    (df["amt"] < lower_limit) |
    (df["amt"] > upper_limit)
).astype(int)
```

```python
print("Total transactions:", len(df))
print("Outliers:", df["amount_outlier"].sum())
print("Outlier percentage:", df["amount_outlier"].mean() * 100)
```

```python
outlier_analysis = df.groupby("amount_outlier").agg(
    total_transactions=("is_fraud", "size"),
    fraudulent_transactions=("is_fraud", "sum"),
    average_amount=("amt", "mean")
)

outlier_analysis["fraud_rate"] = (
    outlier_analysis["fraudulent_transactions"] /
    outlier_analysis["total_transactions"] * 100
)

print(outlier_analysis)
```

```python
outlier_analysis["fraud_rate"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Fraud Rate: Outliers vs Non-Outliers")
plt.xlabel("Outlier Status")
plt.ylabel("Fraud Rate (%)")
plt.xticks(
    [0, 1],
    ["Non-Outliers", "Outliers"],
    rotation=0
)
plt.show()
```

**Geographic Analysis**

**Q.22. How different are customer and merchant locations?**

```python
from math import radians, sin, cos, sqrt, atan2

def calculate_distance(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return 6371 * c

df["distance_km"] = df.apply(
    lambda row: calculate_distance(
        row["lat"], row["long"],
        row["merch_lat"], row["merch_long"]
    ),
    axis=1
)
```
